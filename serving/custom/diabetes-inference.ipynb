# Diabetes Prediction Model Inference

Make inference requests to a diabetes prediction model deployed with a custom Seldon MLServer serving runtime.

The model predicts whether a patient has diabetes based on medical features.

### 1. Import the required libraries.

```python
import requests
```

### 2. Configure the model service endpoint.

```python
MODEL_SERVICE_URL = "http://diabetes-predictor.serving-custom.svc.cluster.local:8080"
MODEL_NAME = "diabetes"
```

### 3. Load the model.

Models deployed from OCI images require explicit loading via the repository API.

```python
load_url = f"{MODEL_SERVICE_URL}/v2/repository/models/{MODEL_NAME}/load"
response = requests.post(load_url)

if response.status_code == 200:
    print(f"✓ Model '{MODEL_NAME}' loaded successfully")
elif response.status_code == 409:
    print(f"✓ Model '{MODEL_NAME}' already loaded")
else:
    print(f"Error: {response.status_code} - {response.text}")
```

### 4. Define patient test cases.

**Features:** Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI, DiabetesPedigreeFunction, Age

```python
patient_1 = {
    "Pregnancies": 6,
    "Glucose": 148,
    "BloodPressure": 72,
    "SkinThickness": 35,
    "Insulin": 0,
    "BMI": 33.6,
    "DiabetesPedigreeFunction": 0.627,
    "Age": 50
}

patient_2 = {
    "Pregnancies": 1,
    "Glucose": 85,
    "BloodPressure": 66,
    "SkinThickness": 29,
    "Insulin": 0,
    "BMI": 26.6,
    "DiabetesPedigreeFunction": 0.351,
    "Age": 31
}
```

### 5. Make predictions.

Send inference requests using the KServe V2 REST protocol.

```python
# Prepare input data
input_data = [
    list(patient_1.values()),
    list(patient_2.values())
]

# Create inference request payload
payload = {
    "inputs": [{
        "name": "predict",
        "shape": [2, 8],
        "datatype": "FP64",
        "data": input_data
    }]
}

# Send request
infer_url = f"{MODEL_SERVICE_URL}/v2/models/{MODEL_NAME}/infer"
response = requests.post(infer_url, json=payload)
response.raise_for_status()

# Parse predictions
result = response.json()
predictions = result["outputs"][0]["data"]

# Display results
classes = ['No diabetes', 'Diabetes']
print("Predictions:")
print(f"  Patient 1: {classes[int(predictions[0])]}")
print(f"  Patient 2: {classes[int(predictions[1])]}")
```
